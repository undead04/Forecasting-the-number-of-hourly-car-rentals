# Chạy toàn pipeline

Chạy các notebook theo thứ tự `01_eda → 02_clean → 03_features → 04_split → 05_train → 06_evaluate`.
Cell dưới thực hiện toàn bộ; hoặc mở từng notebook chạy tay.

```python
import json
from pathlib import Path
ROOT = Path.cwd()
if not (ROOT / "datas").exists() and (ROOT.parent / "datas").exists():
    ROOT = ROOT.parent
SRC = ROOT / "src"
ORDER = ["01_eda.ipynb", "02_clean.ipynb", "03_features.ipynb",
         "04_split.ipynb", "05_train.ipynb", "06_evaluate.ipynb"]
for nb_file in ORDER:
    print(f"\n{'=' * 60}\n>>> {nb_file}\n{'=' * 60}")
    nb = json.loads((SRC / nb_file).read_text(encoding="utf-8"))
    code = "\n\n".join("".join(c["source"]) for c in nb["cells"]
                         if c["cell_type"] == "code")
    exec(compile(code, nb_file, "exec"), {"__name__": "__main__"})
print("\nPipeline HOÀN TẤT. Xem kết quả trong datas/01_eda .. datas/06_evaluation")
```
